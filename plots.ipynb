```python
import matplotlib.pyplot as plt

# Data extracted from the images
n_values = [5, 8, 16, 24]

# Probabilities (%) for each hand type
hands_data = {
    "One Pair": [42.2569, 31.3728, 0, 0],
    "Straight": [0.3925, 8.9129, 0.48, 0.0005], # <0.001 treated as 0.0005
    "Flush": [0.1965, 6.7572, 23.02, 0.067],
    "Full House": [0.1441, 6.0664, 63.01, 34.28]
}

# Define colors for each plot
colors = ['#3498db', '#e67e22', '#2ecc71', '#e74c3c']

# Set up the figure and axes (2x2 grid)
fig, axes = plt.subplots(2, 2, figsize=(14, 12))
fig.suptitle('Poker Hand Probabilities vs. Hand Length (n)', fontsize=18, fontweight='bold')

# Flatten axes for easy iteration
axes_flat = axes.flatten()

for i, (hand_name, probabilities) in enumerate(hands_data.items()):
    ax = axes_flat[i]
    
    # Plotting the line and points
    ax.plot(n_values, probabilities, marker='o', markersize=8, linestyle='-', color=colors[i], linewidth=3)
    
    # Formatting titles and labels
    ax.set_title(f'Probability of {hand_name}', fontsize=15, fontweight='bold')
    ax.set_xlabel('Hand Length (n)', fontsize=12)
    ax.set_ylabel('Probability (%)', fontsize=12)
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.set_xticks(n_values)
    
    # Adjust y-axis limit to prevent numbers hitting the top border
    max_val = max(probabilities)
    if max_val > 0:
        ax.set_ylim(-max_val*0.05, max_val * 1.3) # Give 30% extra space at top
    else:
        ax.set_ylim(-1, 10)

    # Add data labels with white background for better visibility
    for x, y in zip(n_values, probabilities):
        label = f'{y:.2f}%' if y >= 0.01 else (f'{y:.4f}%' if y > 0 else '0%')
        ax.annotate(label, 
                    (x, y), 
                    textcoords="offset points", 
                    xytext=(0, 15), 
                    ha='center', 
                    fontsize=11, 
                    fontweight='bold',
                    bbox=dict(boxstyle="round,pad=0.3", fc="white", ec=colors[i], alpha=0.9))

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('poker_probabilities.png')
```
