static rule

```python
import random
from itertools import combinations
from collections import Counter

# ---- Card representation ----
RANKS = list(range(2, 15))  # 2-14, 14 = Ace
SUITS = ['s', 'h', 'd', 'c']
DECK = [(r, s) for r in RANKS for s in SUITS]
ROYAL_RANKS = {10, 11, 12, 13, 14}  # 10, J, Q, K, A

# ---- Hand evaluation ----

def best_hand_rank(cards):
    best = None
    for combo in combinations(cards, min(5, len(cards))):
        rank = hand_rank(combo)
        if best is None or rank > best:
            best = rank
    return best

def hand_rank(cards):
    ranks = sorted([r for r, s in cards], reverse=True)
    suits = [s for r, s in cards]
    counts = Counter(ranks)
    rank_counts = sorted(counts.values(), reverse=True)

    is_flush = len(set(suits)) == 1
    is_straight = (len(set(ranks)) == 5 and ranks[0] - ranks[4] == 4)

    if set(ranks) == {14, 2, 3, 4, 5}:
        is_straight = True
        ranks = [5, 4, 3, 2, 1]
        counts = Counter(ranks)

    tiebreaker = sorted(ranks, key=lambda r: (counts[r], r), reverse=True)

    if is_straight and is_flush:
        return (8, tiebreaker)
    if rank_counts[0] == 4:
        return (7, tiebreaker)
    if rank_counts[:2] == [3, 2]:
        return (6, tiebreaker)
    if is_flush:
        return (5, tiebreaker)
    if is_straight:
        return (4, tiebreaker)
    if rank_counts[0] == 3:
        return (3, tiebreaker)
    if rank_counts[:2] == [2, 2]:
        return (2, tiebreaker)
    if rank_counts[0] == 2:
        return (1, tiebreaker)
    return (0, tiebreaker)

# ---- Royal flush strategy ----

def get_royal_flush_rule(my_hand):
    """
    Pick 1: target any royal rank (10/J/Q/K/A) across all suits.
    Pick 2+: static rule — any royal rank of the committed suit.
    No need to track which ranks we already have since they're already
    out of the deck and the rule can never land on a card we own.
    """
    if not my_hand:
        return lambda c: c[0] in ROYAL_RANKS

    committed_suit = my_hand[0][1]
    return lambda c, suit=committed_suit: c[1] == suit and c[0] in ROYAL_RANKS

# ---- Single game ----

def play_game_royal_flush(n_picks=5, dealer_fill=8):
    my_hand = []
    dealer_hand = []
    remaining_deck = list(DECK)

    for _ in range(n_picks):
        if not remaining_deck:
            break

        rule_fn = get_royal_flush_rule(my_hand)
        matching = [c for c in remaining_deck if rule_fn(c)]
        if not matching:
            break

        random.shuffle(remaining_deck)
        target_idx = next(i for i, c in enumerate(remaining_deck) if rule_fn(c))

        my_hand.append(remaining_deck[target_idx])
        dealer_hand += remaining_deck[:target_idx]
        remaining_deck = remaining_deck[target_idx + 1:]

    if len(my_hand) < 5:
        return 0, my_hand  # incomplete hand loses

    random.shuffle(remaining_deck)
    needed = max(0, dealer_fill - len(dealer_hand))
    final_dealer = dealer_hand + remaining_deck[:needed]

    win = 1 if best_hand_rank(my_hand) > best_hand_rank(final_dealer) else 0
    return win, my_hand

# ---- Simulate ----

def simulate_royal_flush(n_games=2000, n_picks=5, dealer_fill=8):
    wins = 0
    royal_flush_completions = 0

    for _ in range(n_games):
        win, final_hand = play_game_royal_flush(n_picks, dealer_fill)
        wins += win
        hand_ranks = {r for r, s in final_hand}
        hand_suits = {s for r, s in final_hand}
        if hand_ranks == ROYAL_RANKS and len(hand_suits) == 1:
            royal_flush_completions += 1

    win_rate = wins / n_games
    completion_rate = royal_flush_completions / n_games
    return win_rate, completion_rate

if __name__ == '__main__':
    n_games = 2000
    print(f"Simulating royal flush strategy over {n_games} games...\n")
    win_rate, completion_rate = simulate_royal_flush(n_games=n_games)
    print(f"Royal flush completion rate: {completion_rate:.3f}")
    print(f"Win rate:                    {win_rate:.3f}")
    print(f"Losses despite royal flush:  {completion_rate - win_rate:.3f}")
```
